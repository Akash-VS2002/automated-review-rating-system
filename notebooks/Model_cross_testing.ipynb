```python
import pandas as pd
import joblib
from sklearn.metrics import accuracy_score, classification_report

# -----------------------------
# 1️⃣ Load test datasets
# -----------------------------
bal_test_df = pd.read_csv("Bal_test_dataset.csv")
imb_test_df = pd.read_csv("Imb_test_data.csv")

X_bal_test, y_bal_test = bal_test_df["Review"], bal_test_df["Rating"]
X_imb_test, y_imb_test = imb_test_df["Review"], imb_test_df["Rating"]

# -----------------------------
# 2️⃣ Load saved models and vectorizers
# -----------------------------
# Model_A (Balanced)
model_a = joblib.load("Model_A.pkl")
tfidf_bal = joblib.load("tfidf_bal.pkl")

# Model_B (Imbalanced)
model_b = joblib.load("Model_B.pkl")
tfidf_imb = joblib.load("tfidf_imb.pkl")

# -----------------------------
# 3️⃣ Define a helper function for evaluation
# -----------------------------
def evaluate_model(model, vectorizer, X_test, y_test, name):
    X_test_vec = vectorizer.transform(X_test)
    y_pred = model.predict(X_test_vec)
    acc = accuracy_score(y_test, y_pred)
    report = classification_report(y_test, y_pred)
    print(f"\n=== {name} ===")
    print(f"Accuracy: {acc:.4f}")
    print("Classification Report:")
    print(report)
    return acc

# -----------------------------
# 4️⃣ Cross-test
# -----------------------------
print("🔹 Cross-testing models on both datasets:")

# Model_A on Balanced Test
evaluate_model(model_a, tfidf_bal, X_bal_test, y_bal_test, "Model_A on Balanced Test")

# Model_A on Imbalanced Test
evaluate_model(model_a, tfidf_bal, X_imb_test, y_imb_test, "Model_A on Imbalanced Test")

# Model_B on Imbalanced Test
evaluate_model(model_b, tfidf_imb, X_imb_test, y_imb_test, "Model_B on Imbalanced Test")

# Model_B on Balanced Test
evaluate_model(model_b, tfidf_imb, X_bal_test, y_bal_test, "Model_B on Balanced Test")
```
